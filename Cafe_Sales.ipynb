```python
columns_to_fill = ['Item','Payment Method','Location']
for column_name in columns_to_fill:
  df.loc[df[column_name].isna(),column_name] = 'UNKNOWN'
df.head()
```

```python
cols_to_fill = ['Quantity','Price Per Unit','Total Spent']
for column_name in cols_to_fill:
  df.loc[df[column_name].isna(),column_name] = 0
df.head()
```

```python
col_to_fill = ['Transaction Date']
for column_name in col_to_fill:
  df.loc[df[column_name].isna(),column_name] = '2/21/2023'
df.head()
```

```python
df['Item'] = df['Item'].replace('ERROR','UNKNOWN')
df.head()
```

```python
df.head(15)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df['Total Spent'] = df['Total Spent'].replace('ERROR',np.nan)
df.head()
```

```python
df.info()
```

```python
df['Total Spent'] = df['Total Spent'].fillna(0)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
numerical_cols = ['Quantity','Price Per Unit','Total Spent']
for col in numerical_cols:
  df[col] = pd.to_numeric(df[col],errors='coerce')
df.info()
```

```python
df['Total Spent'] = df['Total Spent'].fillna(0).astype('int64')
df.head()
```

```python
df.info()
```

```python
df['Price Per Unit'] = df['Price Per Unit'].fillna(0).astype('int64')
df.head()
```

```python
df.info()
```

```python
df['Quantity'] = df['Quantity'].fillna(0).astype('int64')
df.head()
```

```python
df.info()
```

```python
type_dict = {
    'Transaction Date':'datetime64[ns]',
    'Item':'category',
    'Payment Method':'category',
    'Location':'category',
    'Quantity':'int64',
    'Price Per Unit':'int64',
    'Total Spent':'int64'
}

df['Transaction Date'] = df['Transaction Date'].replace('ERROR', '2/21/2023')
df['Transaction Date'] = df['Transaction Date'].replace('UNKNOWN', '2/21/2023')
for key, value in type_dict.items():
  df[key] = df[key].astype(value)
df.info()
```
